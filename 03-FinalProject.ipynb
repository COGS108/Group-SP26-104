## Overview

Wearable fitness trackers are widely marketed as tools that improve physical activity, reduce sedentary behavior, and empower users to manage their health. Using data from the Health Information National Trends Survey (HINTS 7), we tested whether wearable use independently predicts four health-related outcomes: weekly exercise, daily sitting time, BMI, and health self-efficacy, after controlling for demographics and health status. The central finding is that wearable use alone explained less than 3% of the variation in any outcome. The factors that actually predicted health behaviors and outcomes were demographics and health status, not device ownership. While wearable use retained a modest association with exercise after controls, it showed no relationship with sitting time or self-efficacy, and wearable users had unexpectedly higher BMI than comparable non-users, suggesting people may adopt trackers in response to weight concerns rather than trackers reducing weight. Taken together, the results suggest that wearable devices are a marker of who is already healthy rather than a driver of better health.

## Research Question

Among U.S. adults, is wearable fitness tracker use independently associated with better health behaviors (physical activity, sedentary time), physical health (BMI), and psychological health confidence (self-efficacy), or is the apparent relationship better accounted for by pre-existing differences in demographics and health status between adopters and non-adopters?

## Background and Prior Work

The proliferation of consumer-grade wearable fitness trackers (devices like the Fitbit, Apple Watch, and Garmin) has introduced a new paradigm in personal health monitoring. These devices are marketed on the premise that tracking physical activity, sedentary behavior, and body metrics will motivate users to exercise more, sit less, and improve their overall health. As a result, wearable technology now occupies a central role in the public health conversation. However, it is unclear whether the association between wearable use and better health reflects a genuine effect of the devices or simply the fact that healthier, wealthier, and more health-conscious people are more likely to buy them in the first place.

A foundational concern is whether wearable devices reliably measure the metrics they claim to track. Research evaluating major commercial trackers found that while heart rate, step count, and sleep duration were measured with relatively high accuracy, energy expenditure showed significantly higher error rates, with mean absolute percentage error reaching as high as 0.44 across devices. For the Fitbit Flex specifically, error in measuring time spent in moderate-to-vigorous physical activity ranged from 7% to 74% depending on the study. These inconsistencies mean that even among adopters, the quality of health feedback varies substantially by device and metric.<sup>1</sup>

The theoretical backbone of this research draws on Bandura's Social Cognitive Theory, particularly the construct of self-efficacy, defined as an individual's belief in their capacity to perform a behavior successfully. Wearable trackers are, in theory, self-efficacy tools: they provide real-time feedback, create visible progress toward goals, and reinforce mastery. Research has shown that trackers can increase participant self-efficacy in supported environments such as wellness groups and physician-led programs. However, a critical and underexamined question is whether wearables improve self-efficacy (the feeling that one can manage their health) without necessarily improving the behavioral or physical outcomes that self-efficacy is supposed to drive. If so, wearables may function more as confidence-builders than as behavior-change instruments.<sup>2</sup>

This distinction between perception and reality has important precedent. Epidemiological research drawing on three nationally representative samples found that individuals who perceived themselves as less active than their peers had a mortality risk up to 72% higher over a 21-year follow-up period, even after controlling for actual physical activity levels. This finding suggests that how people feel about their health can diverge meaningfully from their actual health status, and that tools capable of influencing one without the other deserve careful scrutiny.<sup>3</sup>

An often overlooked dimension of wearable technology is equity. When participant demographics in NIH's 'All of Us' research program were examined, individuals who contributed Fitbit data showed a notable reduction in racial and socioeconomic diversity compared to the broader participant pool, with cost, awareness, and language identified as key barriers. This means that any observed association between wearable use and better health may be confounded by the well-documented relationship between wealth and health. The wearable may be a marker of socioeconomic advantage rather than a cause of better outcomes.<sup>4</sup>

No study using nationally representative survey data has simultaneously tested whether wearable use independently predicts multiple types of health outcomes (behavioral, physical, and psychological) after controlling for the characteristics that distinguish adopters from non-adopters. By comparing the wearable association across exercise, sedentary behavior, BMI, and self-efficacy within a single framework, this project can identify whether wearables relate to some outcomes more than others, and whether the pattern is consistent with genuine device effects or with selection into adoption.

## Hypothesis


We hypothesize that wearable fitness tracker use will be associated with better outcomes across all four measures (more exercise, less sitting, lower BMI, higher self-efficacy) in unadjusted analyses. However, we predict that these associations will weaken substantially after controlling for demographics and health status, suggesting that the wearable-health relationship reflects pre-existing differences between adopters and non-adopters rather than independent effects of the devices themselves.

## Dataset

### Dataset: Health Information National Trends Survey 7 (HINTS 7)

**Link:** https://hints.cancer.gov/data/download-data.aspx  
**Source:** National Cancer Institute  
**Original number of observations:** 7,278 respondents  
**Sampling method:** Stratified mail survey of U.S. adults aged 18+, with oversampling of minority populations  
**Collection period:** 2024  

HINTS 7 is a nationally representative, cross-sectional survey conducted by the National Cancer Institute. It collects information about health communication, health behaviors, healthcare access, technology use, and attitudes toward health among U.S. adults. Each respondent represents a certain number of U.S. adults based on their demographic stratum, captured by survey weights (PERSON_FINWT0). HINTS 7 is appropriate for this project because it contains wearable use alongside multiple health outcome measures and a rich set of demographic and health controls in a single nationally representative sample.

**Outcome variables:**

* WeeklyMinutesModerateExercise — Weekly moderate exercise minutes (behavioral outcome)
* AverageTimeSitting — Average daily hours spent sitting (behavioral outcome)
* BMI — Body mass index, from self-reported height and weight (physical outcome)
* OwnAbilityTakeCareHealth — Health self-efficacy, confidence in managing own health (psychological outcome)

**Predictor of interest:**

* WearableDevTrackHealth2 — Wearable fitness tracker use (current / previous / never)

**Control variables:**

* Age — Respondent age in years
* BirthSex — Sex assigned at birth
* IncomeRanges_IMP — Household income (9 categories, imputed for missing values)
* EducA — Education level (collapsed into 4 categories)
* GeneralHealth — Self-rated general health (5-point scale)
* phq4 — PHQ-4 mental health screener (composite anxiety and depression score)
* PERSON_FINWT0 — Survey weight for population-representative estimates

Note: When a variable is used as an outcome, it is removed from the controls for that model. For example, BMI is a control variable in the exercise models but not in the BMI models where it is the outcome.

### Data Cleaning & Preprocessing

```python
# Import dataset
hints = pd.read_sas('HINTS7_SAS_20250731/hints7_public.sas7bdat')

print("Dataset shape:", hints.shape)
hints.head()
```

```python
# Select variables

hints_clean = hints[[
    'WearableDevTrackHealth2',
    'WeeklyMinutesModerateExercise',
    'AverageTimeSitting',
    'BMI',
    'OwnAbilityTakeCareHealth',
    'GeneralHealth',
    'phq4',
    'Age',
    'BirthSex',
    'IncomeRanges_IMP',
    'EducA',
    'PERSON_FINWT0'
]].copy()

print("Selected dataset shape:", hints_clean.shape)
hints_clean.head()
```

```python
# Rename columns

hints_clean = hints_clean.rename(columns = {
    'WearableDevTrackHealth2': 'Wearable',
    'WeeklyMinutesModerateExercise': 'Exercise_Minutes',
    'AverageTimeSitting': 'Sitting_Hours',
    'BMI': 'BMI',
    'OwnAbilityTakeCareHealth': 'Self_Efficacy',
    'GeneralHealth': 'General_Health',
    'phq4': 'PHQ4',
    'Age': 'Age',
    'BirthSex': 'Sex',
    'IncomeRanges_IMP': 'Income',
    'EducA': 'Education',
    'PERSON_FINWT0': 'Survey_Weight'
})

hints_clean.head()
```

```python
# Check value distributions before cleaning

for col in hints_clean.columns:
    print(f"\n--- {col} ---")
    if hints_clean[col].nunique() <= 15:
        print(hints_clean[col].value_counts().sort_index())
    else:
        print(hints_clean[col].describe())
```

HINTS uses negative values (e.g., -1, -2, -5, -9) to represent missing, skipped, unreadable, or not ascertained responses. These are not meaningful data and must be removed before analysis. For Exercise_Minutes and Sitting_Hours, 0 is a valid response. For PHQ4, 0 is valid (no anxiety or depression symptoms). For all categorical variables, only positive codes represent valid responses.

We use IncomeRanges_IMP (the imputed version) rather than the raw income variable because the HINTS team imputed missing income values using demographic predictors, which reduces the number of rows dropped during cleaning and minimizes potential selection bias from income non-response.

```python
hints_clean = hints_clean[
    (hints_clean['Wearable'] > 0) &
    (hints_clean['Exercise_Minutes'] >= 0) &
    (hints_clean['Sitting_Hours'] >= 0) &
    (hints_clean['BMI'] > 0) &
    (hints_clean['Self_Efficacy'] > 0) &
    (hints_clean['General_Health'] > 0) &
    (hints_clean['PHQ4'] >= 0) &
    (hints_clean['Age'] > 0) &
    (hints_clean['Sex'] > 0) &
    (hints_clean['Income'] > 0) &
    (hints_clean['Education'] > 0) &
    (hints_clean['Survey_Weight'] > 0)
].copy()

print("Cleaned dataset shape:", hints_clean.shape)
print(f"Rows removed: {7278 - hints_clean.shape[0]} ({(7278 - hints_clean.shape[0]) / 7278 * 100:.1f}%)")
```

```python
wearable_map = {1: 'Uses Wearable', 2: 'Used Previously', 3: 'Never Used'}
hints_clean['Wearable_Label'] = hints_clean['Wearable'].map(wearable_map)

wearable_order = ['Never Used', 'Used Previously', 'Uses Wearable']
hints_clean['Wearable_Label'] = pd.Categorical(
    hints_clean['Wearable_Label'], categories = wearable_order, ordered = True
)

print("Wearable use distribution:")
print(hints_clean['Wearable_Label'].value_counts())
```

```python
hints_clean['Self_Efficacy_R'] = 6 - hints_clean['Self_Efficacy']

selfeff_map = {
    5: 'Completely confident',
    4: 'Very confident',
    3: 'Somewhat confident',
    2: 'A little confident',
    1: 'Not confident at all'
}
hints_clean['SelfEfficacy_Label'] = hints_clean['Self_Efficacy_R'].map(selfeff_map)

selfeff_order = [
    'Not confident at all', 'A little confident',
    'Somewhat confident', 'Very confident', 'Completely confident'
]
hints_clean['SelfEfficacy_Label'] = pd.Categorical(
    hints_clean['SelfEfficacy_Label'], categories = selfeff_order, ordered = True
)

print("Self-efficacy distribution (reverse-coded):")
print(hints_clean['SelfEfficacy_Label'].value_counts().sort_index())
```

```python
sex_map = {1: 'Male', 2: 'Female'}
hints_clean['Sex_Label'] = hints_clean['Sex'].map(sex_map)

health_map = {1: 'Excellent', 2: 'Very Good', 3: 'Good', 4: 'Fair', 5: 'Poor'}
hints_clean['Health_Label'] = hints_clean['General_Health'].map(health_map)
health_order = ['Poor', 'Fair', 'Good', 'Very Good', 'Excellent']
hints_clean['Health_Label'] = pd.Categorical(
    hints_clean['Health_Label'], categories = health_order, ordered = True
)

income_map = {
    1: '$0-$9,999', 2: '$10k-$14,999', 3: '$15k-$19,999',
    4: '$20k-$34,999', 5: '$35k-$49,999', 6: '$50k-$74,999',
    7: '$75k-$99,999', 8: '$100k-$199,999', 9: '$200k+'
}
hints_clean['Income_Label'] = hints_clean['Income'].map(income_map)
income_order = list(income_map.values())
hints_clean['Income_Label'] = pd.Categorical(
    hints_clean['Income_Label'], categories = income_order, ordered = True
)

educ_map = {
    1: 'Less than High School', 2: 'High School Graduate',
    3: 'Some College', 4: 'College Graduate or More'
}
hints_clean['Education_Label'] = hints_clean['Education'].map(educ_map)
education_order = list(educ_map.values())
hints_clean['Education_Label'] = pd.Categorical(
    hints_clean['Education_Label'], categories = education_order, ordered = True
)
```

```python
hints_clean['Log_Exercise'] = np.log1p(hints_clean['Exercise_Minutes'])
hints_clean['Exercise_Capped'] = hints_clean['Exercise_Minutes'].clip(upper = 1000)
print("Final dataset shape:", hints_clean.shape)
print("\nMissing values:")
print(hints_clean[['Wearable', 'Exercise_Minutes', 'Sitting_Hours', 'BMI',
                    'Self_Efficacy_R', 'General_Health', 'PHQ4', 'Age',
                    'Sex', 'Income', 'Education', 'Survey_Weight']].isna().sum())

print("\nOutcome variable summaries:")
print(hints_clean[['Exercise_Minutes', 'Sitting_Hours', 'BMI', 'Self_Efficacy_R']].describe().round(2))
```

After cleaning, the dataset retains 6,015 respondents. Exercise_Minutes is strongly right-skewed, with some respondents reporting over 1,000 weekly minutes of moderate exercise, over two hours per day, which may reflect reporting error or non-standard exercise patterns. We use log-transformed exercise in the regression models to reduce the influence of these extreme values and to better satisfy the normality assumptions of linear regression. BMI and Sitting_Hours show moderate right tails but are less severely skewed. Self-efficacy is left-skewed, with most respondents reporting high confidence.

### Regression Analysis

To test whether wearable use independently predicts each health outcome, we fit hierarchical weighted least squares (WLS) regression models using survey weights (PERSON_FINWT0) for population-representative estimates. For each of the four outcomes, we build three nested models:

- **Model 1**: Wearable use only (the unadjusted association)
- **Model 2**: Add demographics (Age, Sex, Income, Education)
- **Model 3**: Add health status controls (the specific controls vary by outcome: each outcome is removed from its own control set)

The logic of the hierarchy: if the wearable coefficient shrinks substantially from Model 1 to Model 3, the unadjusted association was partly or fully confounded by who adopts wearables. If it remains stable, the association is independent of the controls included. Comparing the final wearable coefficient across the four outcomes reveals whether wearables relate to some dimensions of health more than others.

Exercise minutes are log-transformed (using log1p to handle zeros) because the raw distribution is strongly right-skewed. The other three outcomes are modeled on their original scales.

#### Outcome 1: Weekly Exercise Minutes (Log-Transformed)

```python
ex_model1 = smf.wls(
    'Log_Exercise ~ C(Wearable_Label)',
    data = hints_clean, weights = hints_clean['Survey_Weight']
).fit()

ex_model2 = smf.wls(
    'Log_Exercise ~ C(Wearable_Label) + Age + C(Sex_Label) + C(Income_Label) + C(Education_Label)',
    data = hints_clean, weights = hints_clean['Survey_Weight']
).fit()

ex_model3 = smf.wls(
    'Log_Exercise ~ C(Wearable_Label) + Age + C(Sex_Label) + C(Income_Label) + C(Education_Label) + C(Health_Label) + BMI + PHQ4 + Self_Efficacy_R',
    data = hints_clean, weights = hints_clean['Survey_Weight']
).fit()

print("Exercise Models - R-squared:")
print(f"  Model 1 (Wearable only): {ex_model1.rsquared:.4f}")
print(f"  Model 2 (+ Demographics): {ex_model2.rsquared:.4f}")
print(f"  Model 3 (+ Health status): {ex_model3.rsquared:.4f}")
print(f"\nWearable coefficient (Uses Wearable):")
print(f"  Model 1: {ex_model1.params['C(Wearable_Label)[T.Uses Wearable]']:.4f} (p = {ex_model1.pvalues['C(Wearable_Label)[T.Uses Wearable]']:.4f})")
print(f"  Model 2: {ex_model2.params['C(Wearable_Label)[T.Uses Wearable]']:.4f} (p = {ex_model2.pvalues['C(Wearable_Label)[T.Uses Wearable]']:.4f})")
print(f"  Model 3: {ex_model3.params['C(Wearable_Label)[T.Uses Wearable]']:.4f} (p = {ex_model3.pvalues['C(Wearable_Label)[T.Uses Wearable]']:.4f})")
```

```python
outcomes = ['Exercise (log)', 'Sitting Hours', 'BMI', 'Self-Efficacy']
model1_coefs = []
model3_coefs = []

for outcome in outcomes:
    subset = comparison[comparison['Outcome'] == outcome]
    m1 = subset[subset['Model'] == 'Wearable Only']['Wearable_Coef'].values[0]
    m3 = subset[subset['Model'] == '+ Health Status']['Wearable_Coef'].values[0]
    model1_coefs.append(m1)
    model3_coefs.append(m3)

pct_retained = [abs(m3 / m1) * 100 if m1 != 0 else 0 for m1, m3 in zip(model1_coefs, model3_coefs)]

retention_df = pd.DataFrame({
    'Outcome': outcomes,
    'Pct_Retained': pct_retained
})

plt.figure(figsize = (10, 6))
sns.barplot(data = retention_df, x = 'Outcome', y = 'Pct_Retained')
plt.axhline(y = 100, color = 'red', linestyle = '--', alpha = 0.5, label = '100% = no change')
plt.title('How Much of the Wearable Effect Survives Controls?')
plt.xlabel('Health Outcome')
plt.ylabel('% of Unadjusted Coefficient Retained After Controls')
plt.legend()
plt.tight_layout()
plt.show()

print("Coefficient retention:")
for i, row in retention_df.iterrows():
    print(f"  {row['Outcome']}: {row['Pct_Retained']:.1f}% of original coefficient retained")
```

The cross-outcome comparison reveals the central finding of this project. The wearable-health relationship is not uniform, it depends on which outcome you measure.

**Exercise** retained about 76% of its wearable coefficient after controls, and the association remained statistically significant (p < 0.001). This is the one outcome where wearable use shows a persistent independent relationship. However, the effect is modest: the full model explains about 13.7% of the variance in log exercise, and wearable use alone explains only 2.4%, meaning that demographics and health status account for far more variation than wearable ownership.

**Sitting hours** showed no significant association with wearable use at any stage (p > 0.50 throughout). Despite many wearables featuring sedentary reminders and "stand up" alerts, wearing a tracker has no detectable relationship with how much time people spend sitting.

**BMI** produced the most unexpected result. The wearable coefficient was positive (wearable users had higher BMI) and actually increased after adding controls from 0.83 to 1.31 BMI points. This is inconsistent with the idea that wearables help reduce body weight. One plausible interpretation is reverse causation: individuals with higher BMI may be more likely to adopt wearables specifically as a weight management tool. Alternatively, the demographic controls (particularly income and education, which are negatively associated with BMI) may have been masking the positive BMI-wearable association in the unadjusted model. The cross-sectional design cannot determine which explanation is correct.

**Self-efficacy** showed a small initial association with wearable use, but this association shrank to near zero and became non-significant (p = 0.14) after adding health status controls. This means that wearable users do not report higher confidence in managing their health once you compare people with the same general health, BMI, and mental health scores. The apparent self-efficacy advantage of wearable users was driven by their better health status, not by the device itself.

## Ethics

The data we used, HINTS 7, is a publicly available, de-identified dataset from the National Cancer Institute. No names, addresses, or personal identifiers were accessible to us at any point.

A few ethical considerations are worth being upfront about. The most important is causation. Because this is a one-time survey, we can only observe associations, not causes. When we say wearable users exercise more, we cannot say the tracker made them do it. They might have been exercising that much before they bought the device. The same applies to BMI going in the unexpected direction: higher-BMI individuals may be buying wearables specifically because they want to lose weight, which would mean weight concerns caused wearable adoption, not the other way around. We have been careful throughout to frame results as associations rather than effects, and to flag wherever the direction of causation is genuinely unclear.

The exercise finding deserves a specific note here. It survived our controls, but three plausible explanations remain: the tracker actually motivates more exercise, already-active people are just more likely to buy trackers, or some unmeasured factor like personality or intrinsic motivation drives both. A single cross-sectional survey cannot distinguish between these, and we do not claim it can. Only a study that follows people over time, before and after they start using a wearable, could settle this.

Everything in HINTS is self-reported, which is a limitation we cannot get around. People misremember how much they exercise, tend to overestimate their height and underestimate their weight, and may answer health questions in ways that reflect how they want to see themselves rather than reality. The PHQ-4 mental health screener is also a brief tool designed for screening, not clinical diagnosis. We treat all of these as approximations rather than ground truth.

We also thought carefully about how to frame the results without implying that people who do not use wearables are making a bad personal choice. The data show that non-adoption tracks with lower income and education. That reflects cost and access barriers, not individual priorities. We tried to keep that structural reality visible throughout our interpretation.

Finally, about 17% of respondents were removed during cleaning due to missing data. If those respondents are systematically different from the ones we kept, that could skew our results in ways we cannot fully detect. We used the pre-imputed income variable to reduce one major source of missingness, and we applied survey weights in all regression models so the estimates reflect the U.S. adult population rather than just the people who completed every question.

## Discussion and Conclusion

The central finding of this project is straightforward: wearable fitness tracker use alone explained less than 3% of the variation in any of the four health outcomes we tested. Exercise, sitting time, BMI, and self-efficacy were all essentially unpredicted by whether someone wore a tracker. The large gains in explanatory power came from demographics and health status, not from the device. Who you are matters far more than what is on your wrist.

We expected the wearable associations to weaken across all four outcomes after controls, and that was partially right. Sitting time and self-efficacy showed no meaningful association with wearable use after adjustment. But exercise held on even after full controls, meaning wearable users still exercised somewhat more than comparable non-users. And BMI went the wrong direction: wearable users had higher BMI after controls, with the gap actually growing rather than shrinking as we added variables. Neither of these findings changes the main conclusion. Even for exercise, the full model only explained about 14% of the variation in total, and the wearable contributed a small fraction of that. The strongest predictors were general health status and mental health, not device ownership.

The individual outcome results are worth noting. The exercise association is genuinely ambiguous, the tracker might motivate more activity, but it could equally reflect that already-active people buy trackers to monitor what they were already doing. Maybe its some mix of both. The sitting result is notable because stand-up reminders are one of the most marketed wearable features, yet we found zero relationship with sedentary time. The BMI finding suggests people with higher BMI may be adopting wearables specifically to lose weight, making wearable use a response to poor health outcomes rather than a solution to them. And self-efficacy's apparent boost disappeared entirely once we accounted for underlying health, meaning wearable users felt more confident because they were already healthier, not because of the device.

There are real limits here. HINTS is a one-time survey so we cannot establish causation in any direction. Everything is self-reported. The wearable variable treats all users identically regardless of how often they use the device or which features they engage with. And HINTS does not capture neighborhood safety, gym access, work schedules, or social support, all of which likely matter for exercise behavior.

But those limitations do not change the core result. Less than 3% of variance explained across four outcomes is not a finding that calls for nuance about which features work best or which populations benefit most. It is a finding that should make anyone skeptical of treating wearable technology as a public health intervention. The structural and psychological factors that actually shape health behavior are more important than any device can address on its own.
