## Overview

Wearable fitness trackers are increasingly promoted as tools for monitoring physical activity and encouraging healthier lifestyles. However, differences in exercise behavior may reflect more than device use alone. In this project, we used data from the Health Information National Trends Survey (HINTS 7) to examine whether physical activity among U.S. adults is better explained by wearable fitness tracker use, health self-efficacy, or socioeconomic factors such as income and education. Our findings suggest that although wearable users reported slightly higher levels of physical activity, wearable use alone was a weak predictor of exercise behavior. Health self-efficacy and socioeconomic status appeared to explain more variation in weekly exercise levels, suggesting that psychological confidence and access-related factors may play a larger role in physical activity than wearable technology itself.

## Background and Prior Work

The proliferation of consumer-grade wearable fitness trackers (devices like the Fitbit, Apple Watch, and Garmin) has introduced a new paradigm in personal health monitoring. Marketed on the promise of allowing users to better understand their levels of physical activity with scrutiny previously inaccessible, they are now closing the gap between how active a person thinks they are compared to the reality of their activity. As such, these devices now occupy a central role in the conversation of public health. However, it is questionable that the relationship between tracker use, self-perception, and behavioral changes are as simple as is provided by the marketing labels placed on such tools, and the benefits communicated are not distributed evenly across socioeconomic groups.

A foundational concern in this area is whether wearable devices reliably measure physical activity in the first place. Research evaluating major commercial trackers found that while metrics like heart rate, step count, and sleep duration were measured with relatively high accuracy, energy expenditure (calories burned) showed significantly higher error rates, with mean absolute percentage error reaching as high as 0.44 across devices. This matters because caloric output is one of the primary metrics users rely on when assessing whether their activity levels are "adequate." A closer analysis of the issue also reinforces the worry. For devices like the Fitbit Flex, the error in measuring time spent in moderate-to-vigorous physical activity ranged dramatically: from 7% to 74% depending on the study. This indicats substantial variability in actual accuracy. These inconsistencies suggest that the alignment between perceived and actual activity a tracker produces may be artifact-dependent, not just behavior-dependent, and this is a genuine threat to the validity that any study in this space must account for.<a name="cite_ref-1"></a><sup>1</sup>

The theoretical backbone of this research draws on Bandura's Social Cognitive Theory (SCT), particularly the construct of self-efficacy. Self-efficacy is defined as an individual's belief in their capacity to perform a behavior successfully. SCT proposes that self-efficacy is the most influential determinant of physical activity behavior, with high self-efficacy leading to stronger expectations, more concrete goals, and greater persistence when faced with obstacles. Wearable trackers are, in theory, powerful self-efficacy tools: they provide real-time feedback, create visible progress toward goals, and reinforce mastery in activities. According to current data, trackers have been shown to increase participant self-efficacy by providing real, visible reminders of commitment to activity and immediate feedback on a person's progression, though it might be noted that some of this evidence comes from intervention studies with supported environments (wellness groups, physician-led talks, etc), which might be seen to put into question whether these effects generalize to everyday use. Nonetheless, it is a reflection of real relationships between opinion, optimism, and generalizable evidence.<a name="cite_ref-2"></a><sup>2</sup>

One of the most interesting but underexplored findings in this literature is that the perception of physical activity may carry independent health significance beyond actual behavior. Epidemiological research drawing on three nationally representative samples found that individuals who perceived themselves as less active than their peers had a mortality risk up to 72% higher over a 21-year follow-up period, even after controlling for actual physical activity levels. This suggests that how accurately one perceives their own activity. It isn't just the activity someone performs, but also their personal thoughts on their well-being that hold real health implications, and that tools capable of correcting or distorting that perception (like the wearables studied in this proposal) deserve serious empirical scrutiny.<a name="cite_ref-3"></a><sup>3</sup>
An often overlooked dimension of wearable technology is equity. That fitness trackers democratize health data is an assumption that is not often considered or supported. When participant demographics in a large national health research program (NIH's 'All of Us') were examined, it was found that individuals who contributed fitbit data showed a notable reduction in racial and socioeconomic diversity as compared to the broader pool of overall participants, with cost, awareness, and language found to be the key barriers when it came to contribution. Additionally, sociological research shows that physical activity trackers are used more frequently from individuals from more affluent backgrounds, which reflect a potential concern that technological health tools may reinforce rather than reduce existing health inequalities. This idea, mostly absent from device-focussed accuracy studies, shows an important alternative view that must be considered in any model that claims to understand the efficacy of health trackers at a national level.<a name="cite_ref-4"></a><sup>4</sup>
The current literature establishes 3 main tensions. For one, trackers are assumed to improve the alignment between perceived activity and actual activity, yet device accuracy varies significantly by metric and context. Second, self-efficacy is a hard to define mediator for behavioral change, though tracker based self-efficacy gains might be context-dependent, and finally, socioeconomic status shapes both tracker access and the conditions in which feedback loops are effective. No study using survey data has simultaneously tied together all three of these factors in a general adult population. This is the gap this proposal will address.

## Hypothesis


We hypothesized that individuals who use wearable fitness trackers would report higher levels of physical activity than non-users. However, we expected this relationship to be moderated by socioeconomic status (SES) and health self-efficacy. Specifically, we anticipated that individuals with higher levels of health self-efficacy and greater socioeconomic resources would report the highest levels of physical activity, regardless of wearable device use. This expectation reflects disparities in technology access, engagement quality, and the social conditions under which health-monitoring tools are most effective. Prior research suggests that the benefits of wearable devices are often most pronounced in supportive, resource-rich environments, while financial barriers, differences in health literacy, and varying levels of confidence in managing one's health may limit meaningful engagement with these technologies. As a result, we predict that self-efficacy and socioeconomic status will explain more variation in physical activity than wearable use alone.

## Data Analysis

### Setup

```python
# Import packages for data cleaning and analysis
import pandas as pd
import numpy as np

# Import packages for data visualization
import matplotlib.pyplot as plt
import seaborn as sns

# Import statistical testing functions
from scipy.stats import f_oneway, spearmanr

# Import regression modeling package
import statsmodels.formula.api as smf

# Set plot style for cleaner visualizations
sns.set(style='whitegrid')

# Make plots display at a readable size
plt.rcParams['figure.figsize'] = (10, 6)

# Display more columns when viewing dataframes
pd.set_option('display.max_columns', None)
```

### Data Cleaning & Preprocessing

```python
# import dataset

hints = pd.read_sas('HINTS7_SAS_20250731/hints7_public.sas7bdat')

print("Dataset shape:", hints.shape)
hints.head()
```

```python
# select variables needed for the project

hints_clean = hints[[
    'WearableDevTrackHealth2',
    'WeeklyMinutesModerateExercise',
    'OwnAbilityTakeCareHealth',
    'IncomeRanges',
    'Education'
]].copy()

print("Selected dataset shape:", hints_clean.shape)
hints_clean.head()
```

```python
# rename columns to make them easier to understand

hints_clean = hints_clean.rename(columns={
    'WearableDevTrackHealth2': 'Wearable',
    'WeeklyMinutesModerateExercise': 'Exercise_Minutes',
    'OwnAbilityTakeCareHealth': 'Self_Efficacy',
    'IncomeRanges': 'Income',
    'Education': 'Education'
})

hints_clean.head()
```

```python
# Check values before cleaning
# Used to identify negative survey response codes that need to be removed

print("Wearable values:")
print(hints_clean['Wearable'].value_counts().sort_index())

print("\nExercise minutes values:")
print(hints_clean['Exercise_Minutes'].describe())

print("\nSelf-efficacy values:")
print(hints_clean['Self_Efficacy'].value_counts().sort_index())

print("\nIncome values:")
print(hints_clean['Income'].value_counts().sort_index())

print("\nEducation values:")
print(hints_clean['Education'].value_counts().sort_index())
```

```python
# Check for standard missing values in the selected variables
# This checks for actual NaN values, but does not catch HINTS negative missing-value codes

hints_clean.isna().sum()
```

HINTS does not represent all missing or invalid responses as regular NaN values. Instead, many missing, skipped, unreadable, or invalid survey responses are stored as negative numeric codes. 

Because of this, simply using dropna() would not fully clean the dataset. We still checked for NaN values as an initial step, but the main cleaning process focused on removing invalid negative survey response codes based on the HINTS codebook.

```python
# Remove invalid HINTS survey response codes
# Negative values represent missing, skipped, unreadable, or not ascertained responses
# 0 is kept for Exercise_Minutes because 0 minutes of exercise is a valid response

hints_clean = hints_clean[
    (hints_clean['Wearable'] > 0) &
    (hints_clean['Exercise_Minutes'] >= 0) &
    (hints_clean['Self_Efficacy'] > 0) &
    (hints_clean['Income'] > 0) &
    (hints_clean['Education'] > 0)
].copy()

print("Cleaned dataset shape:", hints_clean.shape)
hints_clean.head()
```

```python
# Recode wearable fitness tracker use

wearable_map = {
    1: 'Uses Wearable',
    2: 'Used Previously',
    3: 'Never Used'
}

hints_clean['Wearable_Label'] = hints_clean['Wearable'].map(wearable_map)

hints_clean['Wearable_Label'].value_counts()
```

```python
# Recode health self-efficacy

selfeff_map = {
    1: 'Completely confident',
    2: 'Very confident',
    3: 'Somewhat confident',
    4: 'A little confident',
    5: 'Not confident at all'
}

hints_clean['SelfEfficacy_Label'] = hints_clean['Self_Efficacy'].map(selfeff_map)

hints_clean['SelfEfficacy_Label'].value_counts()
```

```python
# Recode income categories

income_map = {
    1: '$0-$9,999',
    2: '$10k-$14,999',
    3: '$15k-$19,999',
    4: '$20k-$34,999',
    5: '$35k-$49,999',
    6: '$50k-$74,999',
    7: '$75k-$99,999',
    8: '$100k-$199,999',
    9: '$200k+'
}

hints_clean['Income_Label'] = hints_clean['Income'].map(income_map)

hints_clean['Income_Label'].value_counts()
```

```python
# Recode education categories

education_map = {
    1: 'Less than 8 years',
    2: '8-11 years',
    3: 'High School',
    4: 'Vocational/Technical',
    5: 'Some College',
    6: 'College Graduate',
    7: 'Postgraduate'
}

hints_clean['Education_Label'] = hints_clean['Education'].map(education_map)

hints_clean['Education_Label'].value_counts()
```

```python
# Set ordered categories for plotting

wearable_order = [
    'Never Used',
    'Used Previously',
    'Uses Wearable'
]

selfeff_order = [
    'Not confident at all',
    'A little confident',
    'Somewhat confident',
    'Very confident',
    'Completely confident'
]

income_order = [
    '$0-$9,999',
    '$10k-$14,999',
    '$15k-$19,999',
    '$20k-$34,999',
    '$35k-$49,999',
    '$50k-$74,999',
    '$75k-$99,999',
    '$100k-$199,999',
    '$200k+'
]

education_order = [
    'Less than 8 years',
    '8-11 years',
    'High School',
    'Vocational/Technical',
    'Some College',
    'College Graduate',
    'Postgraduate'
]

hints_clean['Wearable_Label'] = pd.Categorical(
    hints_clean['Wearable_Label'],
    categories=wearable_order,
    ordered=True
)

hints_clean['SelfEfficacy_Label'] = pd.Categorical(
    hints_clean['SelfEfficacy_Label'],
    categories=selfeff_order,
    ordered=True
)

hints_clean['Income_Label'] = pd.Categorical(
    hints_clean['Income_Label'],
    categories=income_order,
    ordered=True
)

hints_clean['Education_Label'] = pd.Categorical(
    hints_clean['Education_Label'],
    categories=education_order,
    ordered=True
)
```

```python
# Create capped exercise variable for visualization only
# The original Exercise_Minutes column will still be used for statistical analysis

hints_clean['Exercise_Minutes_Capped'] = hints_clean['Exercise_Minutes'].clip(upper=1000)

hints_clean[['Exercise_Minutes', 'Exercise_Minutes_Capped']].describe()
```

```python
# Final checks after cleaning

print("Final cleaned dataset shape:", hints_clean.shape)

print("\nMissing values:")
print(hints_clean.isna().sum())

print("\nFinal columns:")
print(hints_clean.columns.tolist())

hints_clean.head()
```

```python
# Summary statistics for key numeric variables

hints_clean[['Exercise_Minutes', 'Wearable', 'Self_Efficacy', 'Income', 'Education']].describe()
```

```python
# Statistical test for differences in exercise by wearable group

wearable_users = hints_clean.loc[hints_clean['Wearable_Label'] == 'Uses Wearable', 'Exercise_Minutes']
former_users = hints_clean.loc[hints_clean['Wearable_Label'] == 'Used Previously', 'Exercise_Minutes']
never_users = hints_clean.loc[hints_clean['Wearable_Label'] == 'Never Used', 'Exercise_Minutes']

f_stat, p_val = f_oneway(wearable_users, former_users, never_users)

print("ANOVA F-statistic:", f_stat)
print("ANOVA p-value:", p_val)
```

Current wearable users reported somewhat higher weekly exercise minutes than respondents who had never used a wearable device, while former users reported the lowest median exercise. However, the distributions overlap substantially, suggesting that wearable use alone may not strongly explain differences in physical activity. A one-way ANOVA showed a statistically significant but modest difference across groups.

### Statistical Analysis

#### Hierarchical Regression Models

```python
# Fit hierarchical linear regression models
# Outcome variable: weekly moderate exercise minutes

# Model 1: Wearable use only
model1 = smf.ols(
    'Exercise_Minutes ~ C(Wearable_Label)',
    data=hints_clean
).fit()

# Model 2: Wearable use + health self-efficacy
model2 = smf.ols(
    'Exercise_Minutes ~ C(Wearable_Label) + C(SelfEfficacy_Label)',
    data=hints_clean
).fit()

# Model 3: Wearable use + health self-efficacy + socioeconomic status
model3 = smf.ols(
    'Exercise_Minutes ~ C(Wearable_Label) + C(SelfEfficacy_Label) + C(Income_Label) + C(Education_Label)',
    data=hints_clean
).fit()
```

```python
# Compare model fit across the three regression models

model_comparison = pd.DataFrame({
    'Model': [
        'Wearable Only',
        'Wearable + Self-Efficacy',
        'Wearable + Self-Efficacy + SES'
    ],
    'R²': [
        model1.rsquared,
        model2.rsquared,
        model3.rsquared
    ],
    'Adjusted R²': [
        model1.rsquared_adj,
        model2.rsquared_adj,
        model3.rsquared_adj
    ]
})

# Compare the coefficient for current wearable users across the three models

wearable_coef_comparison = pd.DataFrame({
    'Model': [
        'Wearable Only',
        'Wearable + Self-Efficacy',
        'Wearable + Self-Efficacy + SES'
    ],
    'Current Wearable User Coefficient': [
        model1.params['C(Wearable_Label)[T.Uses Wearable]'],
        model2.params['C(Wearable_Label)[T.Uses Wearable]'],
        model3.params['C(Wearable_Label)[T.Uses Wearable]']
    ],
    'p-value': [
        model1.pvalues['C(Wearable_Label)[T.Uses Wearable]'],
        model2.pvalues['C(Wearable_Label)[T.Uses Wearable]'],
        model3.pvalues['C(Wearable_Label)[T.Uses Wearable]']
    ]
})

print("Model Comparison:")
display(model_comparison)

print("Wearable Coefficient Comparison:")
display(wearable_coef_comparison)
```

```python
plt.figure(figsize=(9, 6))

sns.barplot(
    data=model_comparison,
    x='Model',
    y='R²'
)

plt.title('Model Comparison: Variance Explained in Weekly Exercise Minutes')
plt.xlabel('Regression Model')
plt.ylabel('R²')

plt.xticks(rotation=20)
plt.tight_layout()
plt.show()
```

The hierarchical regression models showed that wearable use alone explained very little variation in weekly exercise minutes. After adding health self-efficacy, the model explained more variation, suggesting that confidence in managing one's health was more strongly related to exercise than wearable use alone. Adding income and education further improved the model slightly.

The coefficient for current wearable users became smaller after adding self-efficacy and socioeconomic status. This suggests that part of the relationship between wearable use and exercise may be explained by differences in health confidence, income, and education rather than wearable use by itself.

#### Sensitivity Analysis: Log-Transformed Exercise Minutes

```python
# Create a log-transformed version of weekly exercise minutes
# log1p is used because some respondents reported 0 exercise minutes

hints_clean['Log_Exercise_Minutes'] = np.log1p(hints_clean['Exercise_Minutes'])

# Fit the same hierarchical models using log-transformed exercise minutes

log_model1 = smf.ols(
    'Log_Exercise_Minutes ~ C(Wearable_Label)',
    data=hints_clean
).fit()

log_model2 = smf.ols(
    'Log_Exercise_Minutes ~ C(Wearable_Label) + C(SelfEfficacy_Label)',
    data=hints_clean
).fit()

log_model3 = smf.ols(
    'Log_Exercise_Minutes ~ C(Wearable_Label) + C(SelfEfficacy_Label) + C(Income_Label) + C(Education_Label)',
    data=hints_clean
).fit()

# Compare model fit for log-transformed models

log_model_comparison = pd.DataFrame({
    'Model': [
        'Wearable Only',
        'Wearable + Self-Efficacy',
        'Wearable + Self-Efficacy + SES'
    ],
    'R²': [
        log_model1.rsquared,
        log_model2.rsquared,
        log_model3.rsquared
    ],
    'Adjusted R²': [
        log_model1.rsquared_adj,
        log_model2.rsquared_adj,
        log_model3.rsquared_adj
    ]
})

# Compare wearable coefficients for log-transformed models

log_wearable_coef_comparison = pd.DataFrame({
    'Model': [
        'Wearable Only',
        'Wearable + Self-Efficacy',
        'Wearable + Self-Efficacy + SES'
    ],
    'Current Wearable User Coefficient': [
        log_model1.params['C(Wearable_Label)[T.Uses Wearable]'],
        log_model2.params['C(Wearable_Label)[T.Uses Wearable]'],
        log_model3.params['C(Wearable_Label)[T.Uses Wearable]']
    ],
    'p-value': [
        log_model1.pvalues['C(Wearable_Label)[T.Uses Wearable]'],
        log_model2.pvalues['C(Wearable_Label)[T.Uses Wearable]'],
        log_model3.pvalues['C(Wearable_Label)[T.Uses Wearable]']
    ]
})

print("Log-Transformed Model Comparison:")
display(log_model_comparison)

print("Log-Transformed Wearable Coefficient Comparison:")
display(log_wearable_coef_comparison)
```

Because weekly exercise minutes were strongly right-skewed, we repeated the regression analysis using log-transformed exercise minutes. The log-transformed models explained more variation than the raw-minute models. Wearable use alone explained about 2.4% of variation, while the full model including self-efficacy, income, and education explained about 9.1%.

Current wearable use remained statistically significant in the log-transformed models, but its coefficient decreased after adding self-efficacy and socioeconomic status. This supports the conclusion that wearable use is associated with exercise, but part of this relationship is explained by psychological and socioeconomic differences between wearable users and non-users.

## Ethics

This project used the public-use HINTS 7 dataset from the National Cancer Institute. Because the dataset is publicly available and de-identified, we did not have access to names, addresses, or direct personal identifiers. To protect privacy, we analyzed only aggregate patterns across groups and did not attempt to identify individual respondents.

Several ethical concerns were considered. First, because HINTS is observational and cross-sectional, we cannot claim that wearable fitness trackers cause people to exercise more. We therefore framed our results as associations rather than causal effects. Second, the data are self-reported, so exercise minutes, income, wearable use, and self-efficacy may be affected by recall error or social desirability bias.

We also considered the risk of blaming individuals for lower physical activity. Wearable use and exercise may both be shaped by income, education, time, health status, neighborhood safety, and access to exercise spaces. To address this, we included income and education in our models and interpreted wearable use as part of a broader psychological and socioeconomic pattern rather than as a simple individual choice.


## Discussion and Conclusion

This project examined whether wearable fitness tracker use was associated with weekly physical activity among U.S. adults, and whether that relationship was better explained by health self-efficacy and socioeconomic status. Using HINTS 7 data, we analyzed wearable use, weekly moderate exercise minutes, confidence in managing one's health, income, and education.

The exploratory analysis showed that current wearable users reported somewhat higher exercise minutes than respondents who had never used a wearable device. However, the differences across wearable-use groups were modest, and the distributions overlapped substantially. Health self-efficacy showed a clearer relationship with exercise: respondents who were more confident in managing their health tended to report more weekly exercise. Income was also related to both wearable use and exercise, suggesting that socioeconomic access may partly explain the wearable-exercise relationship.

The regression models supported this broader pattern. In the raw exercise-minute models, wearable use alone explained very little variation in weekly exercise. Adding health self-efficacy improved the model more than wearable use alone, and adding income and education improved it slightly further. In the log-transformed sensitivity analysis, wearable use remained statistically significant, but its coefficient decreased after accounting for self-efficacy and socioeconomic status. This suggests that wearable use is associated with exercise, but the relationship is partly explained by psychological and socioeconomic differences.

There are several limitations. HINTS is cross-sectional, so we cannot determine whether wearable use causes higher exercise or whether more active people are simply more likely to use wearables. The variables are also self-reported, which may introduce recall error or social desirability bias. In addition, exercise behavior is complex, and factors such as age, health status, neighborhood safety, work schedule, and access to exercise spaces were not included in this analysis.

Overall, the findings suggest that wearable fitness trackers are related to physical activity, but they are not the whole story. Health self-efficacy and socioeconomic status help explain exercise behavior and reduce the estimated relationship between wearable use and exercise. Public health efforts should therefore avoid treating wearable technology as a simple solution and should also consider confidence, affordability, access, and structural barriers to physical activity.
